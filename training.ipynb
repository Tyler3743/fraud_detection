```python
import numpy as np
from sklearn.metrics import f1_score

def pick_threshold(y_val, p_val):
    ths = np.linspace(0.01, 0.99, 99)
    f1s = [f1_score(y_val, p_val >= t, pos_label=1) for t in ths]
    return ths[int(np.argmax(f1s))]

def eval_at(y_test, p_test, thr):
    return f1_score(y_test, p_test >= thr, pos_label=1)

thr = pick_threshold(y_val, p_val)      # chỉ dùng val
f1_test = eval_at(y_test, p_test, thr)  # áp nguyên lên test
```
